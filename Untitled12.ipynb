```python

#

!pip install -q pandas matplotlib

import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# 3. Load the Titanic dataset
print("Loading Titanic dataset...")

# Look for train.csv in the local directory
if os.path.exists("train.csv"):
    csv_file = "train.csv"
elif os.path.exists("train (1).csv"):
    csv_file = "train (1).csv"
else:
    # Fallback to searching recursively if it was uploaded elsewhere
    csv_files = glob.glob("**/train.csv", recursive=True)
    if not csv_files:
        raise FileNotFoundError("train.csv was not found. Please upload it via the files panel.")
    csv_file = csv_files[0]

df = pd.read_csv(csv_file)

print(f"Dataset loaded successfully from {csv_file}!")
print(df.head())

# 4. Prepare the data
# Pclass: 1 = First Class, 2 = Second Class, 3 = Third Class
# Survived: 0 = Did Not Survive, 1 = Survived

survival_data = (
    df.groupby(["Pclass", "Survived"])
      .size()
      .unstack(fill_value=0)
)

survival_data = survival_data.reindex(
    index=[1, 2, 3],
    columns=[0, 1],
    fill_value=0
)

survival_data.index = [
    "First Class",
    "Second Class",
    "Third Class"
]

survival_data.columns = [
    "Did Not Survive",
    "Survived"
]

print("\nPassenger Survival Counts:")
print(survival_data)

# Create output folder
os.makedirs("chart_outputs", exist_ok=True)

# ============================================================
# 5. BEFORE: POORLY FORMATTED CHART
# ============================================================

fig, ax = plt.subplots(figsize=(11, 7))

# Dark background
fig.patch.set_facecolor("#252525")
ax.set_facecolor("#333333")

x = range(len(survival_data))
width = 0.35

# Too many unrelated colors
colors_bad = [
    "#FF00FF", "#00FFFF",
    "#FFFF00", "#FF4500",
    "#7FFF00", "#9400D3"
]

for i, category in enumerate(survival_data.columns):
    ax.bar(
        [p + i * width for p in x],
        survival_data[category],
        width=width,
        color=colors_bad[i::2],
        label=category
    )

# Poor formatting
ax.set_title(
    "TITANIC DATA",
    color="yellow",
    fontsize=12
)

ax.set_xlabel("CLASS", color="white", fontsize=9)
ax.set_ylabel("NUMBER", color="white", fontsize=9)

ax.set_xticks([p + width / 2 for p in x])
ax.set_xticklabels(
    survival_data.index,
    color="white",
    fontsize=8
)

ax.tick_params(axis="y", colors="white")

# Excessive gridlines
ax.grid(
    True,
    color="white",
    linestyle="--",
    linewidth=1.2
)

ax.legend(
    facecolor="#555555",
    labelcolor="white"
)

plt.tight_layout()

plt.savefig(
    "chart_outputs/before_redesign.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 6. AFTER: IMPROVED CHART
# ============================================================

fig, ax = plt.subplots(figsize=(11, 7))

# Clean background
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# Accessible, consistent colors
colors_good = [
    "#35658D",  # Blue: Did not survive
    "#E69F00"   # Orange: Survived
]

bars_width = 0.35

for i, category in enumerate(survival_data.columns):

    positions = [
        p + i * bars_width
        for p in x
    ]

    bars = ax.bar(
        positions,
        survival_data[category],
        width=bars_width,
        color=colors_good[i],
        label=category,
        edgecolor="white",
        linewidth=0.8
    )

    # Add numerical data labels
    ax.bar_label(
        bars,
        padding=4,
        fontsize=10,
        color="#222222",
        fontweight="bold"
    )

# Clear and descriptive title
ax.set_title(
    "Titanic Passenger Survival by Ticket Class",
    fontsize=16,
    fontweight="bold",
    color="#222222",
    pad=20
)

ax.set_xlabel(
    "Passenger Class",
    fontsize=12,
    fontweight="bold",
    labelpad=10
)

ax.set_ylabel(
    "Number of Passengers",
    fontsize=12,
    fontweight="bold"
)

# Correct positioning
ax.set_xticks([
    p + bars_width / 2
    for p in x
])

ax.set_xticklabels(
    survival_data.index,
    fontsize=11
)

# Minimal horizontal gridlines
ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.set_axisbelow(True)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Clear legend
ax.legend(
    title="Survival Status",
    frameon=False,
    fontsize=10
)

ax.tick_params(axis="both", labelsize=10)

plt.tight_layout()

plt.savefig(
    "chart_outputs/after_redesign.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 7. BEFORE AND AFTER COMPARISON
# ============================================================

print("BEFORE AND AFTER CHARTS CREATED SUCCESSFULLY!")

print("\nBefore Chart (Saved to chart_outputs/before_redesign.png)")
print("\nAfter Chart (Saved to chart_outputs/after_redesign.png)")

# ============================================================
# 8. SAVE DATA SUMMARY
# ============================================================

survival_data.to_csv(
    "chart_outputs/survival_summary.csv"
)

print("\nProject completed successfully!")
print("Charts and summary saved in chart_outputs folder.")
```
